```python
import pandas as pd
import numpy as np
from catboost import CatBoostRegressor
from sklearn.metrics import mean_absolute_error

path_dir = r"C:/Users/Ilya/Downloads/intern-task-regression-courier-deficit-challenge/"

train = pd.read_csv(path_dir + 'train.csv')
test = pd.read_csv(path_dir + 'test.csv')
facts = pd.read_csv(path_dir + 'facts.csv')
shifts = pd.read_csv(path_dir + 'shifts_prediction.csv')


print('Размер train:', train.shape)
print('Размер test:', test.shape)
print('Размер facts:', facts.shape)
print('Размер shifts_prediction:', shifts.shape)
# Проверка названий колонок в исходных таблицах

print(f'Колонки train: \n{train.columns.tolist()}')
print(f'\nКолонки test: \n{test.columns.tolist()}')
print(f'\nКолонки facts: \n{facts.columns.tolist()}')
print(f'\nКолонки shifts_prediction: \n {shifts.columns.tolist()}')
# приведение дат к типу datetime

train['calendar_dt'] = pd.to_datetime(train['calendar_dt'])
facts['calendar_dt'] = pd.to_datetime(facts['calendar_dt'])
shifts['calendar_dt'] = pd.to_datetime(shifts['calendar_dt'])

test['calendar_dt'] = pd.to_datetime('2025-11-24')

print(f'Минимальная дата в train: {train['calendar_dt'].min()}')
print(f'Максимальная дата в train: {train['calendar_dt'].max()}')

print(f'\nМинимальная дата в facts: {facts['calendar_dt'].min()}')
print(f'Максимальная дата в facts: {facts['calendar_dt'].max()}')

print(f'\nМинимальная дата в shifts: {shifts['calendar_dt'].min()}')
print(f'Максимальная дата в shifts: {shifts['calendar_dt'].max()}')

print('\nДата в test:')
print(test['calendar_dt'].unique())
# проверка наличия тестовой даты в таблицах с признаками

test_date = pd.to_datetime('2025-11-24')

print('Есть ли тестовая дата в facts:')
print((facts['calendar_dt'] == test_date).sum())

print('\nЕсть ли тестовая дата в shifts_prediction:')
print((shifts['calendar_dt'] == test_date).sum())

print('\nКоличество store_id в test:')
print(test['store_id'].nunique())
# объединение обучающей выборки с фактическими и прогнозными признаками

train_df = train.merge(
    facts,
    on=['store_id', 'calendar_dt'],
    how='left'
)

train_df = train_df.merge(
    shifts,
    on=['store_id', 'calendar_dt'],
    how='left'
)

# объединение тестовой выборки с фактическими и прогнозными признаками

test_df = test.merge(
    facts,
    on=['store_id', 'calendar_dt'],
    how='left'
)

test_df = test_df.merge(
    shifts,
    on=['store_id', 'calendar_dt'],
    how='left'
)

print('Размер train до объединения:', train.shape)
print('Размер train после объединения:', train_df.shape)

print('\nРазмер test до объединения:', test.shape)
print('Размер test после объединения:', test_df.shape)

display(train_df.head())
display(test_df.head())
# проверка дубликатов по ключам объединения

print('Дубликаты в facts по store_id + calendar_dt:')
print(facts.duplicated(['store_id', 'calendar_dt']).sum())

print('\nДубликаты в shifts_prediction по store_id + calendar_dt:')
print(shifts.duplicated(['store_id', 'calendar_dt']).sum())

print('\nПроверка количества строк:')
print('train_df строк должно быть:', len(train))
print('train_df строк фактически:', len(train_df))

print('\ntest_df строк должно быть:', len(test))
print('test_df строк фактически:', len(test_df))
# проверка пропущенных значений до обработки

print('Пропуски в train_df:')
display(train_df.isna().sum()[train_df.isna().sum() > 0])

print('\nПропуски в test_df:')
display(test_df.isna().sum()[test_df.isna().sum() > 0])
def fill_missing_values(train_df, test_df):

    train_df = train_df.copy()
    test_df = test_df.copy()

    # колонки, где есть пропуски хотя бы в одной из таблиц
    missing_cols = []

    for col in train_df.columns:
        if col in test_df.columns:
            if train_df[col].isna().sum() > 0 or test_df[col].isna().sum() > 0:
                missing_cols.append(col)

    print('Колонки с пропусками:')
    print(missing_cols)

    # добавляем флаги пропусков
    for col in missing_cols:
        train_df[col + '_was_missing'] = train_df[col].isna().astype(int)
        test_df[col + '_was_missing'] = test_df[col].isna().astype(int)

    # предположу что пропуск скорее всего связан с отсутствием маркетинговых расходов
    train_df['marketing_costs_lag_1'] = train_df['marketing_costs_lag_1'].fillna(0)
    test_df['marketing_costs_lag_1'] = test_df['marketing_costs_lag_1'].fillna(0)

    # числовые признаки заполняем медианой по train
    numeric_cols = train_df.select_dtypes(include=['int64', 'float64']).columns.tolist()

    for col in numeric_cols:
        if col == 'target':
            continue

        if col in test_df.columns:
            median_value = train_df[col].median()

            train_df[col] = train_df[col].fillna(median_value)
            test_df[col] = test_df[col].fillna(median_value)

    # категориальные признаки заполняем отдельной категорией
    categorical_cols = train_df.select_dtypes(include=['object']).columns.tolist()

    for col in categorical_cols:
        if col in test_df.columns:
            train_df[col] = train_df[col].fillna('unknown')
            test_df[col] = test_df[col].fillna('unknown')

    return train_df, test_df


train_df, test_df = fill_missing_values(train_df, test_df)
print('Оставшиеся пропуски в train_df:')
display(train_df.isna().sum()[train_df.isna().sum() > 0])

print('\nОставшиеся пропуски в test_df:')
display(test_df.isna().sum()[test_df.isna().sum() > 0])

print('\nРазмер train_df после обработки пропусков:', train_df.shape)
print('Размер test_df после обработки пропусков:', test_df.shape)
def add_features(df):

    df = df.copy()

    df['calendar_dt'] = pd.to_datetime(df['calendar_dt'])

    # календарные признаки
    df['month'] = df['calendar_dt'].dt.month
    df['week'] = df['calendar_dt'].dt.isocalendar().week.astype(int)

    # разница между прогнозной потребностью и числом курьеров, выходивших на смены
    df['staff_gap_by_shifts'] = (
        df['predicted_staff_value'] - df['fact_couriers_with_shifts_lag_1']
    )

    # разница между прогнозной потребностью и числом доступных курьеров
    df['staff_gap_by_available'] = (
        df['predicted_staff_value'] - df['num_available_couriers_lag_1']
    )

    # изменение прогнозного числа заказов относительно прошлой недели
    df['orders_diff'] = (
        df['predicted_num_orders'] - df['fact_num_orders_lag_1']
    )

    # изменение прогнозной нагрузки относительно прошлой недели
    df['load_factor_diff'] = (
        df['predicted_load_factor'] - df['fact_load_factor_lag_1']
    )

    # доля доступных курьеров, которые фактически вышли на смены
    df['shift_participation_rate'] = (
        df['fact_couriers_with_shifts_lag_1'] /
        (df['num_available_couriers_lag_1'] + 1)
    )

    # относительная текучесть курьеров
    df['churn_rate'] = (
        df['fact_staff_churn'] /
        (df['fact_couriers_with_shifts_lag_1'] + 1)
    )

    # прогнозное число заказов на одного необходимого курьера
    df['predicted_orders_per_staff'] = (
        df['predicted_num_orders'] /
        (df['predicted_staff_value'] + 1)
    )

    # прогнозное число заказов на одного доступного курьера
    df['orders_per_available_courier'] = (
        df['predicted_num_orders'] /
        (df['num_available_couriers_lag_1'] + 1)
    )

    # прогнозное число заказов на одного курьера, ранее выходившего на смены
    df['orders_per_shift_courier'] = (
        df['predicted_num_orders'] /
        (df['fact_couriers_with_shifts_lag_1'] + 1)
    )

    # отношение прогнозной потребности к фактической потребности прошлой недели
    df['staff_value_ratio'] = (
        df['predicted_staff_value'] /
        (df['fact_staff_value_lag_1'] + 1)
    )

    # отношение прогнозных заказов к фактическим заказам прошлой недели
    df['orders_ratio'] = (
        df['predicted_num_orders'] /
        (df['fact_num_orders_lag_1'] + 1)
    )

    # отношение прогнозной нагрузки к фактической нагрузке прошлой недели
    df['load_factor_ratio'] = (
        df['predicted_load_factor'] /
        (df['fact_load_factor_lag_1'] + 1)
    )

    # сколько доступных курьеров остаётся после покрытия прогнозной потребности
    df['available_minus_needed'] = (
        df['num_available_couriers_lag_1'] - df['predicted_staff_value']
    )

    # сколько курьеров, ранее выходивших на смены, остаётся после покрытия прогнозной потребности
    df['shifts_minus_needed'] = (
        df['fact_couriers_with_shifts_lag_1'] - df['predicted_staff_value']
    )

    # отношение числа курьеров, выходивших на смены, к прогнозной потребности
    df['shifts_to_predicted_staff_ratio'] = (
        df['fact_couriers_with_shifts_lag_1'] /
        (df['predicted_staff_value'] + 1)
    )

    # отношение доступных курьеров к прогнозной потребности
    df['available_to_predicted_staff_ratio'] = (
        df['num_available_couriers_lag_1'] /
        (df['predicted_staff_value'] + 1)
    )

    return df


train_df = add_features(train_df)
test_df = add_features(test_df)

print(f'Размер train_df после генерации признаков: {train_df.shape}')
print(f'Размер test_df после генерации признаков: {test_df.shape}')

display(train_df.head())
# проверка что после генерации признаков не появились бесконечности или пропуски

numeric_cols = train_df.select_dtypes(include=[np.number]).columns

print('Количество inf в train_df:')
print(np.isinf(train_df[numeric_cols]).sum().sum())

numeric_cols_test = test_df.select_dtypes(include=[np.number]).columns

print('\nКоличество inf в test_df:')
print(np.isinf(test_df[numeric_cols_test]).sum().sum())

print('\nПропуски в train_df после генерации признаков:')
display(train_df.isna().sum()[train_df.isna().sum() > 0])

print('\nПропуски в test_df после генерации признаков:')
display(test_df.isna().sum()[test_df.isna().sum() > 0])
# метрика WAPE

def wape(y_true, y_pred):

    y_true = np.array(y_true)
    y_pred = np.array(y_pred)

    return np.sum(np.abs(y_true - y_pred)) / np.sum(y_true)
# временная валидация по последней доступной дате

last_date = train_df['calendar_dt'].max()

train_part = train_df[train_df['calendar_dt'] < last_date].copy()
valid_part = train_df[train_df['calendar_dt'] == last_date].copy()

print('Дата валидации:', last_date)
print('Размер train_part:', train_part.shape)
print('Размер valid_part:', valid_part.shape)
print('Сумма target на валидации:', valid_part['target'].sum())
print('Средний target на валидации:', valid_part['target'].mean())
print('Доля нулевых target на валидации:', (valid_part['target'] == 0).mean())
# выбор признаков

target = 'target'

drop_cols = [
    'target',
    'calendar_dt'
]

features = [
    col for col in train_df.columns
    if col not in drop_cols
]

cat_features = train_df[features].select_dtypes(include=['object']).columns.tolist()

print('Количество признаков:', len(features))
print('Категориальные признаки:', cat_features)

print('\nПервые 10 признаков:')
print(features[:10])
# обучение базовой модели CatBoost

model = CatBoostRegressor(
    iterations=3000,
    learning_rate=0.03,
    depth=6,
    loss_function='MAE',
    eval_metric='MAE',
    random_seed=42,
    verbose=200,
    early_stopping_rounds=200
)

model.fit(
    train_part[features],
    train_part[target],
    eval_set=(valid_part[features], valid_part[target]),
    cat_features=cat_features
)

valid_pred = model.predict(valid_part[features])

# дефицит курьеров не может быть отрицательным
valid_pred = np.clip(valid_pred, 0, None)

print('Validation WAPE:', wape(valid_part[target], valid_pred))
print('Validation MAE:', mean_absolute_error(valid_part[target], valid_pred))
print('Best iteration:', model.get_best_iteration())
# расчёт важности признаков

feature_importance = pd.DataFrame({
    'feature': features,
    'importance': model.get_feature_importance()
})

feature_importance = feature_importance.sort_values(
    by='importance',
    ascending=False
)

display(feature_importance.head(20))
# обучение финальной модели на всех доступных обучающих данных

best_iteration = model.get_best_iteration()

if best_iteration is None:
    best_iteration = 607

print('Используемое количество итераций:', best_iteration)

final_model = CatBoostRegressor(
    iterations=best_iteration,
    learning_rate=0.03,
    depth=6,
    loss_function='MAE',
    eval_metric='MAE',
    random_seed=42,
    verbose=200
)

final_model.fit(
    train_df[features],
    train_df[target],
    cat_features=cat_features
)
# прогноз на тестовой выборке

test_pred = final_model.predict(test_df[features])

# дефицит курьеров не может быть отрицательным
test_pred = np.clip(test_pred, 0, None)

print('Минимальный прогноз:', test_pred.min())
print('Максимальный прогноз:', test_pred.max())
print('Средний прогноз:', test_pred.mean())
print('Медианный прогноз:', np.median(test_pred))
print('Количество нулевых прогнозов:', (test_pred == 0).sum())
print('Доля нулевых прогнозов:', (test_pred == 0).mean())
# формирование submission-файла
sample_submission = pd.read_csv(path_dir + 'sample_submission.csv')
submission = sample_submission.copy()

print('Колонки sample_submission:')
print(submission.columns.tolist())

if 'target' in submission.columns:
    submission['target'] = test_pred
else:
    submission.iloc[:, -1] = test_pred

display(submission.head())
display(submission.describe())

submission.to_csv(path_dir + 'submission.csv', index=False)

print('Файл сохранён: /content/submission.csv')
print('Размер submission:', submission.shape)
test_pred_rounded = np.round(test_pred)
test_pred_rounded = np.clip(test_pred_rounded, 0, None)

submission_rounded = sample_submission.copy()
submission_rounded['target'] = test_pred_rounded

display(submission_rounded.head())
display(submission_rounded.describe())

submission_rounded.to_csv(path_dir + 'submission_final.csv', index=False)

print('Файл сохранён')
model_rmse = CatBoostRegressor(
    iterations=3000,
    learning_rate=0.03,
    depth=6,
    loss_function='RMSE',
    eval_metric='MAE',
    random_seed=52,
    verbose=200,
    early_stopping_rounds=200
)

model_rmse.fit(
    train_part[features],
    train_part[target],
    eval_set=(valid_part[features], valid_part[target]),
    cat_features=cat_features
)

valid_pred_rmse = model_rmse.predict(valid_part[features])
valid_pred_rmse = np.clip(valid_pred_rmse, 0, None)

print('RMSE model Validation WAPE:', wape(valid_part[target], valid_pred_rmse))
print('RMSE model Validation MAE:', mean_absolute_error(valid_part[target], valid_pred_rmse))
print('RMSE model Best iteration:', model_rmse.get_best_iteration())
valid_pred_ensemble_90_10 = 0.9 * valid_pred + 0.1 * valid_pred_rmse
valid_pred_ensemble_80_20 = 0.8 * valid_pred + 0.2 * valid_pred_rmse
valid_pred_ensemble_70_30 = 0.7 * valid_pred + 0.3 * valid_pred_rmse

valid_pred_ensemble_90_10 = np.clip(valid_pred_ensemble_90_10, 0, None)
valid_pred_ensemble_80_20 = np.clip(valid_pred_ensemble_80_20, 0, None)
valid_pred_ensemble_70_30 = np.clip(valid_pred_ensemble_70_30, 0, None)

print('MAE model WAPE:', wape(valid_part[target], valid_pred))
print('RMSE model WAPE:', wape(valid_part[target], valid_pred_rmse))
print('Ensemble 90/10 WAPE:', wape(valid_part[target], valid_pred_ensemble_90_10))
print('Ensemble 80/20 WAPE:', wape(valid_part[target], valid_pred_ensemble_80_20))
print('Ensemble 70/30 WAPE:', wape(valid_part[target], valid_pred_ensemble_70_30))
model_mae_depth5 = CatBoostRegressor(
    iterations=3000,
    learning_rate=0.03,
    depth=5,
    loss_function='MAE',
    eval_metric='MAE',
    random_seed=42,
    verbose=200,
    early_stopping_rounds=200
)

model_mae_depth5.fit(
    train_part[features],
    train_part[target],
    eval_set=(valid_part[features], valid_part[target]),
    cat_features=cat_features
)

valid_pred_depth5 = model_mae_depth5.predict(valid_part[features])
valid_pred_depth5 = np.clip(valid_pred_depth5, 0, None)

print('Base depth 6 WAPE:', wape(valid_part[target], valid_pred))
print('Depth 5 Validation WAPE:', wape(valid_part[target], valid_pred_depth5))
print('Depth 5 Validation MAE:', mean_absolute_error(valid_part[target], valid_pred_depth5))
print('Depth 5 Best iteration:', model_mae_depth5.get_best_iteration())
model_mae_depth7 = CatBoostRegressor(
    iterations=3000,
    learning_rate=0.03,
    depth=7,
    loss_function='MAE',
    eval_metric='MAE',
    random_seed=42,
    verbose=200,
    early_stopping_rounds=200
)

model_mae_depth7.fit(
    train_part[features],
    train_part[target],
    eval_set=(valid_part[features], valid_part[target]),
    cat_features=cat_features
)

valid_pred_depth7 = model_mae_depth7.predict(valid_part[features])
valid_pred_depth7 = np.clip(valid_pred_depth7, 0, None)

print('Base depth 6 WAPE:', wape(valid_part[target], valid_pred))
print('Depth 7 Validation WAPE:', wape(valid_part[target], valid_pred_depth7))
print('Depth 7 Validation MAE:', mean_absolute_error(valid_part[target], valid_pred_depth7))
print('Depth 7 Best iteration:', model_mae_depth7.get_best_iteration())
```
